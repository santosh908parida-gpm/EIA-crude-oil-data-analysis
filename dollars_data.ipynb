```python
import pandas as pd

url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=TWEXBGSMTH"

usd_index = pd.read_csv(url)

usd_index.head()
```

```python
usd_index = usd_index.rename(
    columns={
        "observation_date": "month",
        "TWEXBGSMTH": "usd_index"
    }
)
```

```python
usd_index.head()
```

```python
usd_index["month"] = pd.to_datetime(
    usd_index["month"]
).dt.to_period("M")
```

```python
usd_index.head()
```

```python
usd_index.to_csv("usd_index.csv" , index = False)
```
