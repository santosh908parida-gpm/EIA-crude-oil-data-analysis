```python
df['month'] = pd.to_datetime(df['month'])

df = (
    df.sort_values('month').reset_index(drop= True)
)

df.head()
```

```python
df = df[df['month'] <= "2026-02-01"].copy()
```

##### Forecasting Target Def

```python
df["target_brent_next_month"] = (df["brent_price"].shift(-1))
```

##### Lagged brent price

```python
for lag in [1,2,3,6,12]:
    df["brent_price{lag}"] = (df["brent_price"].shift(lag))
```

##### Market price

```python
market_features = [
    "wti_price",
    "usd_index",
    "opec_crude_production",
    "total_non_opec_production_mbpd",
    "global_oil_consumption",
    "total_US_production",
    "us_crude_inventory",
    "cushing_inventory",
    "refinery_utilization",
    "us_crude_imports",
    "us_crude_exports"
]
```

```python
for col in market_features:

    df[f"{col}_lag1"] = df[col].shift(1)
    df[f"{col}_lag3"] = df[col].shift(3)
```

##### Target column(brent_price) momentum

```python
df["brent_return_1m"] = (
    df["brent_price"].pct_change()
)

df["brent_return_3m"] = (
    df["brent_price"].pct_change(3)
)

df["brent_ma3"] = (
    df["brent_price"]
    .rolling(3)
    .mean()
)

df["brent_ma6"] = (
    df["brent_price"]
    .rolling(6)
    .mean()
)

df_model = df.dropna().copy()    ## NaN values created by lag is removed
```

```python
# X and y def
target = "target_brent_next_month"

drop_columns = [
    "month",
    target
]

X = df_model.drop(
    columns=drop_columns
)

y = df_model[target]
```

```python
test_size = 24

X_train = X.iloc[:-test_size]
X_test  = X.iloc[-test_size:]

y_train = y.iloc[:-test_size]
y_test  = y.iloc[-test_size:]
```

```python
print(X_train.shape)
print(X_test.shape)
```

##### Baseline model

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

baseline_pred = X_test["brent_price"]

baseline_mae = mean_absolute_error(
    y_test,
    baseline_pred
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        baseline_pred
    )
)

print("Baseline MAE :", baseline_mae)
print("Baseline RMSE:", baseline_rmse)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=500,
    max_depth=6,
    min_samples_leaf=3,
    random_state=42
)

rf_model.fit(
    X_train,
    y_train
)

rf_pred = rf_model.predict(
    X_test
)

```

```python
evaluate_model(
    y_test,
    rf_pred
)
```

```python
def evaluate_model(y_true, y_pred):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    }
```

```python
evaluate_model(
    y_test,
    rf_pred
)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = GradientBoostingRegressor(
    n_estimators=300,
    learning_rate=0.03,
    max_depth=2,
    loss="huber",
    random_state=42
)

gb_model.fit(
    X_train,
    y_train
)

gb_pred = gb_model.predict(
    X_test
)

evaluate_model(
    y_test,
    gb_pred
)
```

```python
from xgboost import XGBRegressor
```

```python
xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.03,
    max_depth=3,
    min_child_weight=3,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,
    reg_lambda=1.0,
    objective="reg:squarederror",
    random_state=42
)
```

```python
xgb_model.fit(
    X_train,
    y_train
)
```

```python
xgb_pred = xgb_model.predict(
    X_test
)
```

```python
evaluate_model(
    y_test,
    xgb_pred
)
```

```python
results = pd.DataFrame({

    "Baseline": evaluate_model(
        y_test,
        baseline_pred
    ),

    "Random Forest": evaluate_model(
        y_test,
        rf_pred
    ),

    "Gradient Boosting": evaluate_model(
        y_test,
        gb_pred
    ),

    "XGBoost": evaluate_model(
        y_test,
        xgb_pred
    )

}).T

results.sort_values("RMSE")
```
